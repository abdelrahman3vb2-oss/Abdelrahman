```python
df["year_added"] = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month
df["month_name"] = df["date_added"].dt.month_name()

```

```python
df['date_added'].head()
```

```python
df.isna().sum()
```

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
```

```python
df.isna().sum()
```

```python
df['year_added'].hist(bins=df['year_added'].nunique())
plt.title("Histogram of Content Added by Year")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

```python
most_common_date = df["date_added"].mode()[0]
df["date_added"].fillna(most_common_date, inplace=True)
```

```python
df['rating'] = df['rating'].fillna('Unknown')
```

```python
df.isna().sum()
```

```python
df['year_added'] = df['year_added'].fillna(df['year_added'].mode()[0])
df['month_added'] = df['month_added'].fillna(df['month_added'].mode()[0])
df['month_name'] = df['month_added'].fillna(df['month_added'].mode()[0])
```

```python
df.isna().sum()
```

```python
df['duration'] = df['duration'].fillna('Unknown')
```

```python
df.isna().sum()
```

```python
top_genres = df["listed_in"].value_counts().head(10)
print(top_genres)
```

```python
top_genres.plot(kind="barh", figsize=(10,6), color="orange")
plt.title("Top 10 Genres")
plt.xlabel("Number of Shows/Movies")
plt.ylabel("Genre")
plt.show()
```

```python
df.groupby(["year_added", "director"])["show_id"].count().sort_values(ascending=False).head(10)
```

```python
top_directors = df["director"].value_counts().head(10)
print("Top 10 Directors by number of shows/movies:")
print(top_directors)

```

```python
import matplotlib.pyplot as plt

top_directors.plot(kind="bar", figsize=(10,5), color="skyblue")
plt.title("Top 10 Directors by Number of Shows/Movies")
plt.xlabel("Director")
plt.ylabel("Number of Shows/Movies")
plt.xticks(rotation=45)
plt.show()
```

```python
top_countries = df["country"].value_counts().head(10)
print("Top 10 Countries by Number of Shows/Movies:")
print(top_countries)
```

```python
top_countries.plot(kind="bar", figsize=(10,5), color="purple")
plt.title("Top 10 Countries by Number of Shows/Movies")
plt.xlabel("Country")
plt.ylabel("Number of Shows/Movies")
plt.xticks(rotation=45)
plt.show()
```
